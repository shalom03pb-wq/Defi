# Subir el proyecto a GitHub desde Colab
Necesita: (1) `taller_valoracion_con_datos_v2.zip`, (2) el repositorio `Defi/taller-valoracion` ya creado, (3) un *token* de GitHub.

**Crear el token** (en github.com, con "Sitio para computadora"): foto de perfil → Settings → Developer settings → Personal access tokens → **Tokens (classic)** → Generate new token (classic) → marque solo **repo** → Expiration: 1 day → Generate → copie el token (empieza por `ghp_`). Bórrelo al terminar.

```
from google.colab import files
import zipfile, os, shutil
up = files.upload()
nombre = list(up.keys())[0]
shutil.rmtree("/content/proy", ignore_errors=True)
zipfile.ZipFile(nombre).extractall("/content/proy")
raiz = "/content/proy/taller_valoracion"
shutil.rmtree(raiz + "/.github", ignore_errors=True)   # evita pedir permiso de workflows
print(sorted(os.listdir(raiz)))
```

```
from getpass import getpass
import subprocess
usuario = "Defi"
repo = "taller-valoracion"
token = getpass("Pegue el token de GitHub (no se muestra): ")
os.chdir(raiz)
def sh(c):
    r = subprocess.run(c, shell=True, capture_output=True, text=True)
    print((r.stdout + r.stderr).replace(token, "***")[-1500:])
sh("rm -rf .git && git init -q -b main")
sh('git config user.email "estudiante@example.com" && git config user.name "Equipo taller"')
sh("git add -A && git commit -q -m 'Taller de valoración y riesgo: código, datos de respaldo y resultados'")
sh(f"git remote add origin https://{usuario}:{token}@github.com/{usuario}/{repo}.git")
sh("git push -f -u origin main")
sh("git remote remove origin")
print("Listo. Revise https://github.com/%s/%s" % (usuario, repo))
```
